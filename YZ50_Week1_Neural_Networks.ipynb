```python
import matplotlib.pyplot as plt

x1, x2, x3 = 1.0, 2.0, 3.0
weights = [
    [2, 3, 4],
    [3.4, 1.2, 6],
    [2.2, 8.9, 7]
]
biases = [13, 12, 11]
targets = [1, 0, 1]

def hesapla_mse(weights, biases):
    z1 = x1*weights[0][0] + x2*weights[0][1] + x3*weights[0][2] + biases[0]
    z2 = x1*weights[1][0] + x2*weights[1][1] + x3*weights[1][2] + biases[1]
    z3 = x1*weights[2][0] + x2*weights[2][1] + x3*weights[2][2] + biases[2]
    return ((z1-targets[0])**2 + (z2-targets[1])**2 + (z3-targets[2])**2) / 3


w_degerleri = []
loss_degerleri = []
for w in range(-60, 41):
    weights[0][0] = w
    w_degerleri.append(w)
    loss_degerleri.append(hesapla_mse(weights, biases))
weights[0][0] = 2

plt.plot(w_degerleri, loss_degerleri)
plt.xlabel("weights[0][0]")
plt.ylabel("MSE")
plt.title("weights[0][0] değişirken loss")
plt.show()

```

```python
w = 2
h = 0.0001
learning_rate = 0.1

loss_gecmisi = []
w_gecmisi = []

for i in range(50):
    weights[0][0] = w
    mevcut_loss = hesapla_mse(weights, biases)

    weights[0][0] = w + h
    yeni_loss = hesapla_mse(weights, biases)
    weights[0][0] = w

    gradient = (yeni_loss - mevcut_loss) / h
    w = w - learning_rate * gradient

    loss_gecmisi.append(mevcut_loss)
    w_gecmisi.append(w)

    if i % 10 == 0 or i == 49:
        print(i, "w =", round(w, 3), "loss =", round(mevcut_loss, 2), "gradient =", round(gradient, 3))

plt.figure(figsize=(11, 4))
plt.subplot(1, 2, 1)
plt.plot(loss_gecmisi, marker="o", markersize=3)
plt.xlabel("adım"); plt.ylabel("MSE"); plt.title("gradient descent ile loss")
plt.subplot(1, 2, 2)
plt.plot(w_gecmisi, marker="o", markersize=3)
plt.xlabel("adım"); plt.ylabel("weights[0][0]"); plt.title("w'nin gidişi")
plt.tight_layout(); plt.show()
```
